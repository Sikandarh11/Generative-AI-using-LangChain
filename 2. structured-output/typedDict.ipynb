```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from pydantic import BaseModel, Field
load_dotenv()
```

```python
model = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.9)

class Review(TypedDict):
    key_themes: Annotated[list[str], "Write down all the key themes mentioned in the review."]
    summary: str
    sentiment : Annotated[Literal["pos", "neg"], "Determine if the overall sentiment of the review is positive (pos) or negative (neg)."]
    pros: Annotated[Optional[List[str]], "List the pros mentioned in the review, if any."]
    cons: Annotated[Optional[List[str]], "List the cons mentioned in the review, if any."]
    name: Annotated[Optional[str], "Name of the reviewer, if mentioned."]
structured_model = model.with_structured_output(Review)
result = structured_model.invoke("""I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.

The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, vibrant images even in low light. Zooming up to 100x actually works well for distant objects, but anything beyond 30x loses quality.

However, the weight and size make it a bit uncomfortable for one-handed use. Also, Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already provides? The $1,300 price tag is also a hard pill to swallow.

Pros:
Insanely powerful processor (great for gaming and productivity)
Stunning 200MP camera with incredible zoom capabilities
Long battery life with fast charging
S-Pen support is unique and useful
                                 
Review by Siku
""")
print(result['name'])
```

```python
print(type(result))
print("summary:", result["summary"])
print("sentiment:", result["sentiment"])
print("key_themes:", result["key_themes"])
print("pros:", result["pros"])
print("cons:", result["cons"])
```

```python

from pydantic import BaseModel, Field, EmailStr

class Student(BaseModel):
    name: str = "Siku"
    age: Optional[int] =None
    email: EmailStr 
    cgpa: float = Field(gt=0, lt=4.0, default=0.0, description="CGPA must be between 0 and 4.0 decimal value")
new_student = {"age": 20, "email": "siku@example.com", "cgpa": 3.5}#{"name": "Alice"}
student = Student(**new_student)
print(student)
```

```python
model = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.9)

class Review(BaseModel):
    key_themes: list[str] = Field(description="write down all the key themes mentioned in the review.")
    summary: str = Field(description="Write a concise summary of the review, capturing the main points and overall impression.")
    
    sentiment : Literal["pos", "neg"] = Field(description="Determine if the overall sentiment of the review is positive (pos) or negative (neg).")
    pros: list[str] = Field(description="List the pros mentioned in the review, if any.")
    cons: list[str] = Field(description="List the cons mentioned in the review, if any.")
    name: Optional[str] = Field(description="Name of the reviewer, if mentioned.")


structured_model = model.with_structured_output(Review)
result = structured_model.invoke("""I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.

The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, vibrant images even in low light. Zooming up to 100x actually works well for distant objects, but anything beyond 30x loses quality.

However, the weight and size make it a bit uncomfortable for one-handed use. Also, Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already provides? The $1,300 price tag is also a hard pill to swallow.

Pros:
Insanely powerful processor (great for gaming and productivity)
Stunning 200MP camera with incredible zoom capabilities
Long battery life with fast charging
S-Pen support is unique and useful
                                 
Review by Siku
""")
print(result.name)
```
