```python
import random
class NakliLLM:
    def __init__(self):
        print("LLM Created")

    def predict(self, prompt=None):
        response_list = [
            "The product quality is terrible and I am very dissatisfied with my purchase.",
            "The product quality is excellent and I am very satisfied with my purchase."
        ]
        return {"response": random.choice(response_list)}
llm = NakliLLM()
print(llm.predict())
```

```python
class NakliPromptTemplate:

  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def format(self, input_dict):
    return self.template.format(**input_dict)
template = NakliPromptTemplate(
    template="Classify the sentiment of the following feedback text into positive or negative: \n{feedback}",
    input_variables=["feedback"]
)
template.format({"feedback": "The product quality is terrible and I am very dissatisfied with my purchase."})
```
